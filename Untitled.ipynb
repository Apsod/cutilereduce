```python
import cutilereduce.experimental as exp
from cutilereduce.experimental import *


spec = FoldSpec(
    inputs={
        "query": TensorSpec(("h", "l", "g", "dqk"), "bfloat16"),
        "key": TensorSpec(("h", "r", "dqk"), "bfloat16"),
        "value": TensorSpec(("h", "r", "dv"), "bfloat16"),
    },
    batch_axes=("h", "l", "g"),
    fold_axes=("r",),
)


def attention_map(ctx, inputs):
    scores = ctx.contract(
        inputs["query"],
        inputs["key"],
        over=("dqk",),
        dtype="float32",
    )
    return StructuredValue(
        {
            "log_weight": scores,
            "value": ctx.cast(inputs["value"], "float32"),
        }
    )

def payload_get(ctx, semantic):
    del ctx
    return StructuredValue({
        "z": semantic["log_weight"],
        "payload": semantic["value"]
    })

maps = FoldMaps(
        to_factor = attention_map,
        to_output = payload_get
        )

op = FoldOperator(
        spec,
        LogWeightedSum(('dv',)),
        maps=maps,
)
sizes = {
        'h': 2,
        'l': 2048,
        'g': 4,
        'dqk': 128,
        'r': 2048,
        'dv': 128,
        }

program = op.lower()
space = build_schedule_space(program)
forward = next(schedule for schedule in space.forward if schedule.name == 'online')
specialized = specialize_cutile_schedule(
    program, forward, sizes, tiles={'l': 32, 'r': 64}
)

print(emit_cutile_source(specialized))
```
